# Feature Pipeline Check

English / 中文双语说明：
- This notebook runs the preprocessing + feature construction pipeline once to make sure the data is ready for model training.
- 本 Notebook 用来串联预处理与特征构造流程，确认数据在训练前已经清理妥当。

```python
import pandas as pd
from Model_Train.data_preprocess import prepare_panel_data
from Model_Train.features import FeatureConfig, build_feature_set_x1, build_feature_set_x2, build_feature_set_x3, assemble_feature_matrix
```

```python
DATA_DIR = '.'  # 根目录包含所有 CSV
PATTERN = '*.csv'
FEATURE_SET = 'x2'  # 可切换 'x1'/'x2'/'x3'
FEATURE_CONFIG = FeatureConfig(lookback_months=12)
```

```python
panel = prepare_panel_data(data_dir=DATA_DIR, pattern=PATTERN)
panel.head()
```

```python
if FEATURE_SET == 'x1':
    feature_df = build_feature_set_x1(panel, FEATURE_CONFIG)
elif FEATURE_SET == 'x2':
    feature_df = build_feature_set_x2(panel, FEATURE_CONFIG)
elif FEATURE_SET == 'x3':
    feature_df = build_feature_set_x3(panel, FEATURE_CONFIG)
else:
    raise ValueError('Unsupported FEATURE_SET')
feature_df.head()
```

```python
x, y, ids, dates = assemble_feature_matrix(feature_df)
print('Feature matrix shape:', x.shape)
print('Target vector shape:', y.shape)
print('Unique PERMNOs:', ids.nunique())
print('Date span:', dates.min(), '->', dates.max())
```
